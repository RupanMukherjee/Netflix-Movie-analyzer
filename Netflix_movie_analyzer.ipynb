```python
df['director'] = df['director'].fillna('unknown')
```

```python
df['cast'] = df['cast'].fillna('not available')
```

```python
df['country'] = df['country'].fillna(df['country'].mode()[0])
```

```python
df = df.dropna(subset=['date_added']).copy()
```

```python
df.loc[:,'rating'] = df['rating'].fillna(df['rating'].mode()[0])
```

```python
df = df.dropna(subset=['duration']).copy()
```

```python
#Making chart
warnings.filterwarnings('ignore')
average_country = df['country'].value_counts().head(10)
average_country.plot(kind = 'bar')
plt.title('aveage movies per country')
plt.xlabel('country')
plt.ylabel('movies')
plt.savefig('charts/Average_movies per country_graph')
plt.show()
```

```python
#Release year analysis
df
```

```python
#Highest number of movies in a particular year
df['release_year'].value_counts().head(1)
```

```python
#Release year trend line graph
average_year = df['release_year'].value_counts().sort_index()
average_year.plot(kind = 'line', colormap = 'rainbow')
plt.title('Release year trend')
plt.xlabel('years')
plt.ylabel('movie_counts')
plt.yticks(rotation = 90)
plt.show()
```

```python
#Trend analysis after 2015
year_count = df['release_year'].value_counts().sort_index()
year_count[year_count.index >= 2015].plot(kind = 'line', color = ('red'))
plt.title('Trend analysis after 2015')
plt.xlabel('Release_Years')
plt.ylabel('Movie_Counts')
plt.savefig('charts/Movie_trend_analysis_graph.png')
plt.show()
```

```python
#Oldest Movie
df[df['release_year'] == df['release_year']].min()
```

```python
#Most common genres
Genres = df['listed_in'].str.split(', ')
Genres = Genres.explode()
Genres.value_counts().head(10)
```

```python
#Making of graph
genre = Genres.value_counts().head(10)
genre.plot(kind = 'bar', color = ('red'))
plt.title('Most Popular Genres')
plt.xlabel('Genres')
plt.ylabel('Genres_count')
plt.savefig('charts/Popular_genres.png')
plt.show()
```

```python
#Finding number of comedies 
int(Genres.value_counts()['Comedies'])
```

```python
#Comedies vs Drama graph
comedies_count = int(Genres.value_counts()['Comedies'])
drama_count = int(Genres.value_counts()['Dramas'])
comedy_drama_count = {
    'comedies': comedies_count,
    'Dramas': drama_count
}
pd.Series(comedy_drama_count).plot(kind = 'bar')
plt.title('Comedies vs Drama comparison')
plt.xlabel('Genres')
plt.ylabel('movie_count')
plt.savefig('charts/comedy_drama_comparisongraph')
plt.show()
```

```python
#Finding longes movie
movies = df[df['type'] == 'Movie']
movies['minute'] = movies['duration'].str.replace(' min','').astype(int)
movies.nlargest(1,'minute')[['title','duration']]
```

```python
#Finding shortest movie
movies.nsmallest(1,'minute')[['title','duration']]
```

```python
#Histogram chart of movie duration
movies['minute'].plot(kind = 'hist', bins = 30)
plt.title('Movie_Duration histogram chart')
plt.xlabel('Duration in minutes')
plt.ylabel('Number of movies')
plt.savefig('charts/Movie duration histogram chart.png')
plt.show()
```

```python
#Average movie duration
avg_duration = int(movies['minute'].mean())
print(f'average movie duration is {avg_duration} minutes')

```
